# 📚 Books Dataset Preprocessing Pipeline

```text
final_books_dataset.csv

          ↓

          ↓
Create Target
          ↓
Rating 1-6  → Target 0
Rating 7-10 → Target 1
          ↓
Handle Age
          ↓
Handle Publication Year
          ↓
Clean Text Columns
          ↓
Remove Image URLs
          ↓

          ↓
books_final_preprocessed.csv

````python
print("Ratings:", ratings.shape)
print("Users:", users.shape)
print("Books:", df_books.shape)

print("\nRatings columns:")
print(ratings.columns.tolist())

print("\nUsers columns:")
print(users.columns.tolist())

print("\nBooks columns:")
print(df_books.columns.tolist())
````

````python
# ==============================
# CLEAN COLUMN NAMES
# ==============================

def clean_columns(df):
    df = df.copy()
    df.columns = (
        df.columns.astype(str)
        .str.replace('"', '', regex=False)
        .str.replace("'", "", regex=False)
        .str.strip()
    )
    return df

ratings = clean_columns(ratings)
users = clean_columns(users)
df_books = clean_columns(df_books)


# ==============================
# CLEAN JOIN COLUMNS
# ==============================

ratings["User-ID"] = ratings["User-ID"].astype(str).str.strip()
users["User-ID"] = users["User-ID"].astype(str).str.strip()

ratings["ISBN"] = ratings["ISBN"].astype(str).str.strip()
df_books["ISBN"] = df_books["ISBN"].astype(str).str.strip()


# ==============================
# CLEAN BOOK RATINGS
# ==============================

ratings["Book-Rating"] = (
    ratings["Book-Rating"]
    .astype(str)
    .str.replace('"', '', regex=False)
    .str.strip()
)

ratings["Book-Rating"] = pd.to_numeric(
    ratings["Book-Rating"],
    errors="coerce"
)

# Remove missing and zero ratings
ratings_clean = ratings[
    ratings["Book-Rating"].notna() &
    (ratings["Book-Rating"] > 0)
].copy()

print("Ratings after cleaning:", ratings_clean.shape)


# ==============================
# CREATE TARGET
# ==============================

ratings_clean["Target"] = (
    ratings_clean["Book-Rating"] >= 7
).astype(int)


# ==============================
# REMOVE DUPLICATE USERS
# ==============================

users_clean = users.drop_duplicates(
    subset="User-ID",
    keep="first"
).copy()


# ==============================
# REMOVE DUPLICATE BOOKS
# ==============================

books_clean = df_books.drop_duplicates(
    subset="ISBN",
    keep="first"
).copy()


# ==============================
# MERGE RATINGS + USERS
# ==============================

data = pd.merge(
    ratings_clean,
    users_clean,
    on="User-ID",
    how="inner",
    validate="many_to_one"
)

print("After Ratings + Users:", data.shape)


# ==============================
# MERGE WITH BOOKS
# ==============================

data = pd.merge(
    data,
    books_clean,
    on="ISBN",
    how="inner",
    validate="many_to_one"
)

print("Final data shape:", data.shape)

# Display data
display(data.head(20))
````

## Classification Model Training

- Import the required **Machine Learning models** and evaluation metrics.
- Create and train a **Decision Tree Classifier**.
- Create and train a **Random Forest Classifier** with 100 trees.
- Create and train an **SVM (Support Vector Machine)** model.
- Create and train a **KNN (K-Nearest Neighbors)** model with 5 neighbors.
- Use `X_train_scaled` and `y_train` to train all models.
- Use `X_test_scaled` to make predictions on the test data.
- Compare all five models:
  - Logistic Regression
  - Decision Tree
  - Random Forest
  - SVM
  - KNN
- Calculate the following evaluation metrics:
  - **Accuracy** → Overall correct predictions.
  - **Precision** → Correct positive predictions.
  - **Recall** → Correctly identified positive cases.
  - **F1 Score** → Balance between precision and recall.
- Store the results in a DataFrame called **`results`**.
- Finally, print the model comparison table.

````python
import pandas as pd

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


# 1. Decision Tree

dt = DecisionTreeClassifier(random_state=42)

dt.fit(X_train_scaled, y_train)

y_pred_dt = dt.predict(X_test_scaled)



# 2. Random Forest

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train_scaled, y_train)

y_pred_rf = rf.predict(X_test_scaled)

# 3. SVM


svm = SVC()

svm.fit(X_train_scaled, y_train)

y_pred_svm = svm.predict(X_test_scaled)

# 4. KNN

knn = KNeighborsClassifier(n_neighbors=5)

knn.fit(X_train_scaled, y_train)

y_pred_knn = knn.predict(X_test_scaled)

# 5. Model Comparison

results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "SVM",
        "KNN"
    ],

    "Accuracy": [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_svm),
        accuracy_score(y_test, y_pred_knn)
    ],

    "Precision": [
        precision_score(y_test, y_pred_lr, zero_division=0),
        precision_score(y_test, y_pred_dt, zero_division=0),
        precision_score(y_test, y_pred_rf, zero_division=0),
        precision_score(y_test, y_pred_svm, zero_division=0),
        precision_score(y_test, y_pred_knn, zero_division=0)
    ],

    "Recall": [
        recall_score(y_test, y_pred_lr, zero_division=0),
        recall_score(y_test, y_pred_dt, zero_division=0),
        recall_score(y_test, y_pred_rf, zero_division=0),
        recall_score(y_test, y_pred_svm, zero_division=0),
        recall_score(y_test, y_pred_knn, zero_division=0)
    ],

    "F1 Score": [
        f1_score(y_test, y_pred_lr, zero_division=0),
        f1_score(y_test, y_pred_dt, zero_division=0),
        f1_score(y_test, y_pred_rf, zero_division=0),
        f1_score(y_test, y_pred_svm, zero_division=0),
        f1_score(y_test, y_pred_knn, zero_division=0)
    ]
})

print(results)
````
